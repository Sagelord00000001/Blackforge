```python
# Validate Python environment
import blackforge
from blackforge.core.config import load_config
from blackforge.core.logging import setup_logging
from blackforge.runtime.bootstrap import bootstrap

print("All core imports successful.")
```

```python
# Initialize Blackforge with defaults
app = bootstrap()
status = app.verify()
for k, v in status.items():
    symbol = "PASS" if v else "FAIL"
    print(f"  [{symbol}] {k}")

assert app.healthy(), "Blackforge bootstrap verification failed"
print("\nBlackforge is healthy.")
```
